# MineMentor: Data Science Analysis & Model Evaluation

This notebook demonstrates exploratory data analysis (EDA), dataset distributions, ML classification benchmarks, and retrieval evaluation for **MineMentor**.

```python
import json
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np

# Load summary statistics
with open('../data/processed/pipeline_summary.json', 'r') as f:
    summary = json.load(f)

print(f"Raw dataset records:     {summary['raw_total']:,}")
print(f"Cleaned dataset records: {summary['cleaned_total']:,}")
print(f"Train samples:           {summary['train_count']:,}")
print(f"Test samples:            {summary['test_count']:,}")
```

## 1. Intent Category Distribution
Distribution of clean user inquiries across 16 core Minecraft gameplay intents.

```python
intent_df = pd.Series(summary['intent_distribution']).sort_values(ascending=True)
plt.figure(figsize=(10, 6))
intent_df.plot(kind='barh', color='#2e7d32')
plt.title('Inquiries by Gameplay Intent')
plt.xlabel('Question Count')
plt.show()
```

## 2. Difficulty Distribution
Breakdown by player skill tier (Beginner, Intermediate, Advanced, Expert).

```python
diff_df = pd.Series(summary['difficulty_distribution'])
plt.figure(figsize=(7, 4))
diff_df.plot(kind='bar', color=['#4caf50', '#2196f3', '#ff9800', '#f44336'])
plt.title('Questions by Difficulty Tier')
plt.ylabel('Question Count')
plt.show()
```

## 3. Machine Learning Model Benchmark Comparison
Comparison of the Intent Classifier (16 classes) and Difficulty Classifier (4 classes) on the held-out test split of 26,426 samples.

```python
with open('../data/processed/intent_metrics.json', 'r') as f:
    im = json.load(f)
with open('../data/processed/difficulty_metrics.json', 'r') as f:
    dm = json.load(f)

results = pd.DataFrame({
    'Model': ['Intent Classifier', 'Difficulty Classifier'],
    'Accuracy': [im['accuracy'], dm['accuracy']],
    'Weighted F1': [im['weighted_f1'], dm['weighted_f1']],
    'Macro F1': [im['macro_f1'], dm['macro_f1']]
})
print(results.to_markdown(index=False))
```

Performance on the 13 challenging curated expert test cases.
